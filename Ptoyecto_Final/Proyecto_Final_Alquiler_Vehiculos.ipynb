# 1. Analisis de la Situacion

### 1.1 Contexto del Problema y Justificacion

#### Descripcion del Escenario Operativo
La empresa de alquiler de vehiculos se enfrenta a un escenario de crecimiento operativo acelerado que sobrepasa sus capacidades de gestion manuales o semi-automatizadas. Actualmente, el flujo operativo depende de procesos descentralizados, lo que genera inconsistencias criticas en el manejo de la informacion transaccional y de la flota.

#### Problematicas Principales Identificadas

* **Inconsistencia en Reservas y Disponibilidad:** Falta de validacion en tiempo real que causa sobre-reservas (*overbooking*) o la asignacion de vehiculos no disponibles (por estar en mantenimiento preventivo o en transito entre sucursales).
* **Descontrol en la Flota Diversificada:** Incapacidad de gestionar eficientemente las especificaciones unicas de cada segmento de vehiculo (ej. tiempos e infraestructura de carga para vehiculos electricos, limites de tonelaje para vans de carga y especificaciones tecnicas para automoviles particulares).
* **Gestion Deficiente de Clientes y Beneficios:** Dificultad para segmentar adecuadamente el comportamiento de clientes ocasionales frente a empresariales, impidiendo aplicar reglas de negocio para descuentos corporativos o programas de referidos (`RECOMIENDA`).
* **Falta de Trazabilidad Contractual y Mantenimiento:** Perdida de seguimiento a los contratos activos, seguros vinculados y registros operativos de los talleres internos de cada sucursal.

#### Objetivo del Proyecto
Disenar e implementar un sistema de base de datos relacional modelado en tercera forma normal (3FN) y soportado por un diagrama E-R extendido. Este sistema centraliza la gestion de flotas, controla la disponibilidad en tiempo real mediante la integracion de la agregacion `AlquilerEfectivo` y garantiza la integridad referencial en todo el ciclo de vida del alquiler.

---

### 1.2 Tendencias Actuales del Sector

El sector del alquiler de vehiculos atraviesa una transformacion tecnologica orientada a la automatizacion y la optimizacion operativa. Las principales tendencias tecnologicas con impacto directo en el diseno de bases de datos son:

* **Electrificacion e Integracion Tecnologica de Flotas:** La inclusion de vehiculos electricos requiere el registro y monitoreo de variables criticas como el nivel de bateria, autonomia restante, infraestructura de carga y tiempos de recarga estimados (atributos integrados en la entidad `VEHICULO_ELECTRICO`).
* **Telemetria y Monitoreo IoT en Tiempo Real:** El uso de dispositivos GPS y sensores a bordo genera un flujo constante de datos sobre kilometraje, estado del motor y ubicacion geografica, lo que demanda un historial estructurado de mantenimientos en `TALLER`.
* **Modelos Flexibles de Movilidad y Suscripcion:** La transicion del alquiler tradicional hacia esquemas de suscripcion mensual o corporativa exige esquemas de segmentacion de clientes (`CLIENTE_EMPRESARIAL` vs `CLIENTE_OCASIONAL`) y manejo de beneficios o descuentos parametrizables.
* **Precios Dinamicos y Contratacion Digital:** La automatizacion de contratos y polizas de seguro vinculadas a la reserva requiere estructuras relacionales de tipo N:M (como `CONTRATO` e `INCLUYE` con `POLIZA_SEGURO`) para adaptar cobros segun demanda y nivel de cobertura.

### Descripcion Estructural del Modelo

* **Jerarquias de Especializacion (Total y Disjunta):**
  * **CLIENTE:** Se especializa disjuntamente en `CLIENTE OCASIONAL` (atributos especificos: Tarjeta de Credito, Licencia de Conduccion) y `CLIENTE EMPRESARIAL` (Razon Social, Descuento Corporativo, Contacto Ejecutivo).
  * **VEHICULO:** Se especializa en `AUTOMOVIL` (Numero de Pasajeros, Tipo de Carroceria), `VAN_CARGA` (Capacidad de Carga en Kg, Volumen Util en M3, Categoria de Licencia requerida), `VEHICULO_ELECTRICO` (Autonomia en Km, Tiempo de Carga en Horas) y `VEHICULO_PREMIUM` (Torque, Aceleracion de 0 a 100 km/h, Detalles de Lujo).

* **Entidad Debil:**
  * `TALLER_INTERNO` se modelo inicialmente como entidad debil dependiente de la entidad fuerte `SUCURSAL` mediante la relacion `POSEE` (1:N), utilizando `Cod_Taller` como discriminador parcial (concepto que en la fase relacional se evoluciono hacia talleres independientes para dar soporte a redes externas).

* **Mecanismo de Agregacion (Bloque Operativo Central):**
  * La transaccion central del sistema consolida una relacion ternaria `AlquilerEfectivo` entre `CLIENTE` (1:N), `VEHICULO` (1:N) y `RESERVA` (1:1).
  * Dicho grupo esta delimitado dentro del **Marco de Agregacion**, del cual deriva la relacion 1:1 `FORMALIZA EN` conectada a la entidad `CONTRATO`. Esto garantiza que ningun contrato se genere sin la presencia fisica validada de la reserva, el cliente y el vehiculo entregado.

* **Relaciones Adicionales y Reglas de Negocio:**
  * **Relacion Reflexiva:** `RECOMIENDA` sobre `CLIENTE` con roles definidos (0:N Recomendador / 0:1 Recomendado), modelando programas de referidos comerciales de forma limpia e independiente.
  * **Cobertura de Seguros:** La entidad `CONTRATO` se vincula con `POLIZA_SEGURO` para especificar las coberturas contratadas, la aseguradora y el valor diario del amparo.

# 3. Modelo Relacional y su Normalizacion

### 3.1 Paso de Modelo E-R a Modelo Relacional Inicial

Al realizar la transformacion del Modelo Entidad-Relacion Extendido al Modelo Relacional, se aplicaron mejoras tecnicas orientadas a la viabilidad operativa en entornos reales:

1. **Clave Primaria de Vehiculos:** Se establecio el `VIN` (*Vehicle Identification Number*) como Clave Primaria (PK) en la superclase `VEHICULO`. La placa es un identificador de transito susceptible a cambios administrativos o reasignaciones, mientras que el VIN es inmutable legal y fisicamente.
2. **Aislamiento de Atributos Especializados:** Para evitar valores nulos masivos en vehiculos de combustion tradicional, los atributos exclusivos de carga electrica se aislaron en la subclase `VEHICULO_ELECTRICO`.
3. **Clave Subrogada en Alquiler:** La agregacion ternaria central `AlquilerEfectivo` une Cliente, Reserva y Vehiculo. En lugar de arrastrar una clave compuesta masiva, se definio la clave artificial `IdAlquiler`, optimizando la conexion con `CONTRATO`.
4. **Logistica de Talleres y Mantenimiento:** Se desacoplo el `TALLER` de la `SUCURSAL` para permitir el trabajo con talleres externos aliados o concesionarios. Ademas, el mantenimiento se separo en cabecera (`MANTENIMIENTO`) y detalle (`DETALLE_MANTENIMIENTO`) para discriminar consumos de repuestos y mano de obra.
5. **Control de Reservas:** Se conecto la tabla `RESERVA` directamente con el `CLIENTE` mediante `NIT_Cliente`, asegurando la trazabilidad de la intencion de alquiler antes de que se materialice la entrega del automotor.

### 3.2 Proceso de Normalizacion (1FN, 2FN, 3FN)

<p align="center">
  <img src="../Modelo Relacional/Modelo Relacional Normalizado.png" alt="Modelo Relacional Normalizado" width="1000">
</p>

#### Paso 1: Primera Forma Normal (1FN)
* **Regla Tecnica:** Una tabla esta en 1FN si y solo si todos sus atributos son atomicos; es decir, no contiene campos compuestos ni grupos repetitivos o multivaluados.
* **Diagnostico en el Modelo Inicial:** La entidad `CLIENTE` violaba la 1FN en dos campos:
  1. `Direccion_Compuesta`: Almacenaba simultaneamente ciudad, barrio y calle en una sola cadena. Se soluciono dividiendo el campo en tres columnas atomicas: `Ciudad`, `Barrio` y `Calle`.
  2. `Telefonos`: Almacenaba una coleccion variable de telefonos para un mismo cliente, rompiendo la atomicidad. Se resolvio creando la tabla dependiente `CLIENTE_TELEFONO` con clave primaria compuesta `(NIT, Telefono)`.

#### Paso 2: Segunda Forma Normal (2FN)
* **Regla Tecnica:** Una relacion esta en 2FN si cumple 1FN y todos los atributos no clave tienen dependencia funcional completa respecto a la clave primaria. Se analiza especificamente en relaciones con **claves compuestas** para descartar dependencias parciales.
* **Diagnostico en el Modelo:**
  1. `CONTRATA`: Clave primaria `(NIT, NombreUnico)`. Los atributos `FechaInicio` y `FechaFin` representan la vigencia de esa membresia particular tomada por ese cliente especifico. Dependen de ambos componentes de la clave.
  2. `CLIENTE_TELEFONO`: Clave primaria `(NIT, Telefono)`. Al no poseer atributos descriptivos fuera de la clave, cumple 2FN de forma directa.
  3. Todas las demas tablas poseen claves primarias simples de un solo atributo, por lo que satisfacen la 2FN por definicion.

#### Paso 3: Tercera Forma Normal (3FN)
* **Regla Tecnica:** Una relacion esta en 3FN si cumple 2FN y no contiene dependencias transitivas; ningun atributo no clave puede depender funcionalmente de otro atributo no clave.
* **Diagnostico y Soluciones Aplicadas:**
  1. **Independencia de Talleres:** Si se almacenaran datos como `CiudadTaller` o `DireccionTaller` en `MANTENIMIENTO`, dependerian de `CodTaller` y no directamente de `IdMantenimiento`. La existencia de la tabla independiente `TALLER` asegura que los datos de localizacion dependan unicamente de su clave primaria.
  2. **Especializacion de Flota:** Incluir especificaciones como `AutonomiaKm` o `VolumenUtilM3` en `VEHICULO` crearia inconsistencias y celdas con valores nulos para modelos a gasolina o automoviles estandar. Las tablas hijas `AUTOMOVIL`, `VAN_CARGA`, `VEHICULO_ELECTRICO` y `VEHICULO_PREMIUM` garantizan que cada atributo especializado dependa de forma directa y no transitiva de su `VIN`.
  3. **Agregacion de Alquiler:** El uso de `ALQUILER_EFECTIVO` con clave subrogada `IdAlquiler` desacopla al `CONTRATO` y las `POLIZA_SEGURO` de los detalles internos del cliente y del vehiculo, eliminando transitividades en la facturacion.

# 4. Base de Datos Construida en SQL (Esquema DDL)

Para la implementacion fisica del modelo relacional se selecciono el motor **SQLite**, un sistema de gestion de bases de datos relacionales embebido y robusto que cumple con los estandares SQL ANSI.

Un aspecto tecnico fundamental en SQLite es que el soporte para claves foraneas viene desactivado por defecto para mantener retrocompatibilidad historica. Por esta razon, en la primera celda de ejecucion se activa explicitamente la directiva:
```sql
PRAGMA foreign_keys = ON;
```
Esto garantiza que cualquier intento de violar la integridad referencial (por ejemplo, insertar un vehiculo con una sucursal inexistente o asociar un contrato a un alquiler invalido) sea bloqueado inmediatamente por el motor.

A continuacion, se ejecuta la creacion de la base de datos `alquiler_vehiculos.db` y las sentencias DDL correspondientes a las 19 tablas normalizadas.


````python
import sqlite3
import pandas as pd
import os

# Conexion a la base de datos local SQLite
db_path = "alquiler_vehiculos.db"
conn = sqlite3.connect(db_path)

# Activacion obligatoria de integridad referencial de claves foraneas
conn.execute("PRAGMA foreign_keys = ON;")
cursor = conn.cursor()

# Sentencias DDL para las 19 tablas del esquema definitivo en 3FN
ddl_script = """
-- 1. SUCURSAL
CREATE TABLE IF NOT EXISTS SUCURSAL (
    IdSucursal INTEGER PRIMARY KEY,
    Ciudad VARCHAR(50) NOT NULL,
    Direccion VARCHAR(150) NOT NULL
);

-- 2. TALLER
CREATE TABLE IF NOT EXISTS TALLER (
    CodTaller INTEGER PRIMARY KEY,
    NIT_Taller INTEGER,
    NombreTaller VARCHAR(100) NOT NULL,
    TipoTaller VARCHAR(20) NOT NULL,
    Ciudad VARCHAR(50) NOT NULL,
    Direccion VARCHAR(150) NOT NULL
);

-- 3. MEMBRESIA
CREATE TABLE IF NOT EXISTS MEMBRESIA (
    NombreUnico VARCHAR(50) PRIMARY KEY,
    CostoMensual DECIMAL(10,2) NOT NULL,
    Beneficios TEXT NOT NULL
);

-- 4. CLIENTE
CREATE TABLE IF NOT EXISTS CLIENTE (
    NIT INTEGER PRIMARY KEY,
    CUC VARCHAR(20) UNIQUE NOT NULL,
    Nombre VARCHAR(100) NOT NULL,
    Ciudad VARCHAR(50) NOT NULL,
    Barrio VARCHAR(50) NOT NULL,
    Calle VARCHAR(100) NOT NULL,
    Recomendador_NIT INTEGER,
    FOREIGN KEY (Recomendador_NIT) REFERENCES CLIENTE(NIT)
);

-- 5. CLIENTE_TELEFONO
CREATE TABLE IF NOT EXISTS CLIENTE_TELEFONO (
    NIT INTEGER NOT NULL,
    Telefono VARCHAR(20) NOT NULL,
    PRIMARY KEY (NIT, Telefono),
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

-- 6. CLIENTE_OCASIONAL
CREATE TABLE IF NOT EXISTS CLIENTE_OCASIONAL (
    NIT INTEGER PRIMARY KEY,
    TarjetaCredito VARCHAR(20) NOT NULL,
    LicenciaConduccion VARCHAR(20) NOT NULL,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

-- 7. CLIENTE_EMPRESARIAL
CREATE TABLE IF NOT EXISTS CLIENTE_EMPRESARIAL (
    NIT INTEGER PRIMARY KEY,
    RazonSocial VARCHAR(150) NOT NULL,
    DescuentoCorporativo DECIMAL(5,2) NOT NULL,
    ContactoEjecutivo VARCHAR(100) NOT NULL,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

-- 8. CONTRATA
CREATE TABLE IF NOT EXISTS CONTRATA (
    NIT INTEGER NOT NULL,
    NombreUnico VARCHAR(50) NOT NULL,
    FechaInicio DATE NOT NULL,
    FechaFin DATE NOT NULL,
    PRIMARY KEY (NIT, NombreUnico),
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (NombreUnico) REFERENCES MEMBRESIA(NombreUnico)
);

-- 9. VEHICULO
CREATE TABLE IF NOT EXISTS VEHICULO (
    VIN VARCHAR(17) PRIMARY KEY,
    Placa VARCHAR(10) UNIQUE NOT NULL,
    Marca VARCHAR(50) NOT NULL,
    Modelo VARCHAR(50) NOT NULL,
    Año INTEGER NOT NULL,
    KilometrajeActual INTEGER NOT NULL,
    TarifaDiariaBase DECIMAL(10,2) NOT NULL,
    NumPasajeros INTEGER NOT NULL,
    TipoTransmision VARCHAR(20) NOT NULL,
    TipoTraccion VARCHAR(10) NOT NULL,
    TipoCombustible VARCHAR(20) NOT NULL,
    IdSucursal INTEGER NOT NULL,
    FOREIGN KEY (IdSucursal) REFERENCES SUCURSAL(IdSucursal)
);

-- 10. AUTOMOVIL
CREATE TABLE IF NOT EXISTS AUTOMOVIL (
    VIN VARCHAR(17) PRIMARY KEY,
    TipoCarroceria VARCHAR(30) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

-- 11. VAN_CARGA
CREATE TABLE IF NOT EXISTS VAN_CARGA (
    VIN VARCHAR(17) PRIMARY KEY,
    CapacidadCargaKg DECIMAL(8,2) NOT NULL,
    VolumenUtilM3 DECIMAL(6,2) NOT NULL,
    TipoLicencia VARCHAR(5) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

-- 12. VEHICULO_ELECTRICO
CREATE TABLE IF NOT EXISTS VEHICULO_ELECTRICO (
    VIN VARCHAR(17) PRIMARY KEY,
    AutonomiaKm INTEGER NOT NULL,
    TiempoCargaHoras DECIMAL(4,2) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

-- 13. VEHICULO_PREMIUM
CREATE TABLE IF NOT EXISTS VEHICULO_PREMIUM (
    VIN VARCHAR(17) PRIMARY KEY,
    Torque VARCHAR(30) NOT NULL,
    Aceleracion VARCHAR(30) NOT NULL,
    DetallesLujo TEXT NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

-- 14. RESERVA
CREATE TABLE IF NOT EXISTS RESERVA (
    CodReserva INTEGER PRIMARY KEY,
    NIT_Cliente INTEGER NOT NULL,
    VIN_Solicitado VARCHAR(17),
    FechaCreacion DATE NOT NULL,
    FechaInicio DATE NOT NULL,
    FechaFin DATE NOT NULL,
    Estado VARCHAR(20) NOT NULL,
    FOREIGN KEY (NIT_Cliente) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (VIN_Solicitado) REFERENCES VEHICULO(VIN)
);

-- 15. ALQUILER_EFECTIVO
CREATE TABLE IF NOT EXISTS ALQUILER_EFECTIVO (
    IdAlquiler INTEGER PRIMARY KEY AUTOINCREMENT,
    NIT INTEGER NOT NULL,
    CodReserva INTEGER NOT NULL,
    VIN VARCHAR(17) NOT NULL,
    FechaHoraSalida DATETIME NOT NULL,
    NivelCombSalida VARCHAR(20) NOT NULL,
    KmSalida INTEGER NOT NULL,
    FechaHoraLlegada DATETIME,
    NivelCombLlegada VARCHAR(20),
    KmLlegada INTEGER,
    NotasEstado TEXT,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (CodReserva) REFERENCES RESERVA(CodReserva),
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN)
);

-- 16. CONTRATO
CREATE TABLE IF NOT EXISTS CONTRATO (
    NumContrato INTEGER PRIMARY KEY,
    IdAlquiler INTEGER NOT NULL,
    FechaFirma DATE NOT NULL,
    Condiciones TEXT NOT NULL,
    CostoTotalReal DECIMAL(12,2) NOT NULL,
    FOREIGN KEY (IdAlquiler) REFERENCES ALQUILER_EFECTIVO(IdAlquiler)
);

-- 17. POLIZA_SEGURO
CREATE TABLE IF NOT EXISTS POLIZA_SEGURO (
    CodPoliza INTEGER PRIMARY KEY,
    NumContrato INTEGER NOT NULL,
    Aseguradora VARCHAR(100) NOT NULL,
    TipoCobertura VARCHAR(50) NOT NULL,
    CostoDiario DECIMAL(10,2) NOT NULL,
    FOREIGN KEY (NumContrato) REFERENCES CONTRATO(NumContrato)
);

-- 18. MANTENIMIENTO
CREATE TABLE IF NOT EXISTS MANTENIMIENTO (
    IdMantenimiento INTEGER PRIMARY KEY,
    VIN VARCHAR(17) NOT NULL,
    CodTaller INTEGER NOT NULL,
    FechaIngreso DATE NOT NULL,
    CostoTotalMant DECIMAL(12,2) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN),
    FOREIGN KEY (CodTaller) REFERENCES TALLER(CodTaller)
);

-- 19. DETALLE_MANTENIMIENTO
CREATE TABLE IF NOT EXISTS DETALLE_MANTENIMIENTO (
    IdDetalle INTEGER PRIMARY KEY,
    IdMantenimiento INTEGER NOT NULL,
    RepuestoOServicio VARCHAR(100) NOT NULL,
    Cantidad INTEGER NOT NULL,
    CostoUnitario DECIMAL(10,2) NOT NULL,
    Subtotal DECIMAL(10,2) NOT NULL,
    FOREIGN KEY (IdMantenimiento) REFERENCES MANTENIMIENTO(IdMantenimiento)
);
"""

cursor.executescript(ddl_script)
conn.commit()

# Verificacion de tablas creadas en el catalogo del sistema
tablas = cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';").fetchall()
print(f"Esquema DDL ejecutado correctamente. Total tablas en la base de datos: {len(tablas)}")
for t in sorted([item[0] for item in tablas]):
    print(f"  - {t}")
````

# 6. Consultas Estrategicas sobre la Base de Datos (10 Consultas SQL)

### Consulta 1: Ingresos Totales y Ticket Promedio por Sucursal Comercial
**Problema de Negocio que Responde:**
Permite a la direccion financiera evaluar el desempeno comercial y la rentabilidad por ciudad, identificando cuales sedes estan generando mayor volumen de facturacion neta y cual es el valor promedio percibido por alquiler (*ticket promedio*).

**Aspectos Tecnicos:**
Combina cuatro tablas (`SUCURSAL`, `VEHICULO`, `ALQUILER_EFECTIVO`, `CONTRATO`) mediante `JOIN`, agrupando por sede y calculando agregaciones monetarias con redondeo a dos decimales.


````python
query_1 = """
SELECT
    s.IdSucursal,
    s.Ciudad,
    COUNT(c.NumContrato) AS TotalAlquileres,
    ROUND(SUM(c.CostoTotalReal), 2) AS IngresosTotales,
    ROUND(AVG(c.CostoTotalReal), 2) AS TicketPromedio
FROM SUCURSAL s
JOIN VEHICULO v ON s.IdSucursal = v.IdSucursal
JOIN ALQUILER_EFECTIVO ae ON v.VIN = ae.VIN
JOIN CONTRATO c ON ae.IdAlquiler = c.IdAlquiler
GROUP BY s.IdSucursal, s.Ciudad
ORDER BY IngresosTotales DESC;
"""
df_1 = pd.read_sql_query(query_1, conn)
df_1
````

### Consulta 2: Historial y Costos de Mantenimiento y Repuestos por Vehiculo
**Problema de Negocio que Responde:**
Permite al departamento de taller y flota monitorear el gasto acumulado en reparaciones y repuestos para cada automotor, identificando que taller realizo la labor y cuantos conceptos tecnicos fueron intervenidos.

**Aspectos Tecnicos:**
Realiza una union de cuatro tablas (`VEHICULO`, `MANTENIMIENTO`, `TALLER`, `DETALLE_MANTENIMIENTO`), agrupando por unidad vehicular y calculando el costo total a partir de la suma de subtotales de repuestos.


````python
query_2 = """
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    t.NombreTaller,
    m.FechaIngreso,
    COUNT(dm.IdDetalle) AS TotalItemsAtendidos,
    ROUND(SUM(dm.Subtotal), 2) AS CostoTotalReparaciones
FROM VEHICULO v
JOIN MANTENIMIENTO m ON v.VIN = m.VIN
JOIN TALLER t ON m.CodTaller = t.CodTaller
JOIN DETALLE_MANTENIMIENTO dm ON m.IdMantenimiento = dm.IdMantenimiento
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, t.NombreTaller, m.FechaIngreso
ORDER BY CostoTotalReparaciones DESC;
"""
df_2 = pd.read_sql_query(query_2, conn)
df_2
````

Une la subclase `CLIENTE_EMPRESARIAL` con la tabla relacional `CONTRATA` y el catalogo `MEMBRESIA`, filtrando suscripciones vigentes mediante comparacion de fechas.


````python
query_3 = """
SELECT
    ce.NIT,
    ce.RazonSocial,
    ce.DescuentoCorporativo AS PorcentajeDescuento,
    m.NombreUnico AS PlanMembresia,
    m.CostoMensual,
    co.FechaInicio AS VigenciaDesde,
    co.FechaFin AS VigenciaHasta
FROM CLIENTE_EMPRESARIAL ce
JOIN CONTRATA co ON ce.NIT = co.NIT
JOIN MEMBRESIA m ON co.NombreUnico = m.NombreUnico
WHERE DATE('now') BETWEEN co.FechaInicio AND co.FechaFin
   OR co.FechaFin >= '2026-11-01'
ORDER BY ce.DescuentoCorporativo DESC;
"""
df_3 = pd.read_sql_query(query_3, conn)
df_3
````

### Consulta 5: Comparativa de Kilometraje Recorrido en Servicios de Alquiler Concluidos
**Problema de Negocio que Responde:**
Permite al equipo de operaciones calcular la intensidad de uso (kilometros reales transitados) por cada usuario entre la entrega inicial y la recepcion final, detectando desgaste acelerado o posibles cobros por kilometraje excedente.

**Aspectos Tecnicos:**
Calcula una metrica derivada dinamica `(ae.KmLlegada - ae.KmSalida)` sobre `ALQUILER_EFECTIVO`, combinando con `CLIENTE` y `VEHICULO`, filtrando unicamente registros con entrega completada.


````python
query_5 = """
SELECT
    ae.IdAlquiler,
    cl.Nombre AS Cliente,
    v.Placa,
    v.Marca || ' ' || v.Modelo AS Vehiculo,
    ae.KmSalida,
    ae.KmLlegada,
    (ae.KmLlegada - ae.KmSalida) AS KmRecorridos,
    ae.FechaHoraSalida,
    ae.FechaHoraLlegada
FROM ALQUILER_EFECTIVO ae
JOIN CLIENTE cl ON ae.NIT = cl.NIT
JOIN VEHICULO v ON ae.VIN = v.VIN
WHERE ae.KmLlegada IS NOT NULL
ORDER BY KmRecorridos DESC;
"""
df_5 = pd.read_sql_query(query_5, conn)
df_5
````

### Consulta 6: Tasa de Utilizacion y Facturacion Acumulada por Unidad Vehicular
**Problema de Negocio que Responde:**
Informa a la gerencia que vehiculos presentan mayor rotacion y demanda comercial y cuales unidades estan subutilizadas, apoyando decisiones de adquisicion, promocion o desincorporacion de activos.

**Aspectos Tecnicos:**
Aplica `LEFT JOIN` partiendo desde `VEHICULO` hacia `ALQUILER_EFECTIVO` y `CONTRATO` para incluir unidades sin alquileres, utilizando `COALESCE` para manejar nulos contables y ordenando por frecuencia de despacho.


````python
query_6 = """
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    v.TipoCombustible,
    COUNT(ae.IdAlquiler) AS TotalVecesAlquilado,
    COALESCE(SUM(c.CostoTotalReal), 0) AS TotalFacturadoVehiculo
FROM VEHICULO v
LEFT JOIN ALQUILER_EFECTIVO ae ON v.VIN = ae.VIN
LEFT JOIN CONTRATO c ON ae.IdAlquiler = c.IdAlquiler
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, v.TipoCombustible
ORDER BY TotalVecesAlquilado DESC, TotalFacturadoVehiculo DESC;
"""
df_6 = pd.read_sql_query(query_6, conn)
df_6
````

### Consulta 7: Red de Fidelizacion y Clientes Referidos Mediante Relacion Reflexiva
**Problema de Negocio que Responde:**
Evalua la efectividad del programa de recomendacion boca a boca (`RECOMIENDA`), listando que usuarios han ejercido como recomendadores efectivos y detallando los nombres de sus referidos.

**Aspectos Tecnicos:**
Implementa un auto-join (`SELF JOIN`) sobre la tabla `CLIENTE` utilizando la clave foranea reflexiva `Recomendador_NIT` y la funcion agregada `GROUP_CONCAT` para condensar la lista de clientes atraidos.


````python
query_7 = """
SELECT
    p.NIT AS NIT_Recomendador,
    p.Nombre AS NombreRecomendador,
    COUNT(h.NIT) AS TotalClientesReferidos,
    GROUP_CONCAT(h.Nombre, ', ') AS ClientesRecomendados
FROM CLIENTE p
JOIN CLIENTE h ON p.NIT = h.Recomendador_NIT
GROUP BY p.NIT, p.Nombre
ORDER BY TotalClientesReferidos DESC;
"""
df_7 = pd.read_sql_query(query_7, conn)
df_7
````

### Consulta 8: Cobertura de Polizas de Seguro por Aseguradora y Costo Promedio Diario
**Problema de Negocio que Responde:**
Permite al comite de riesgos y compras analizar la distribucion de polizas emitidas, la participacion de cada aseguradora aliada y el impacto economico del seguro dentro de los contratos formalizados.

**Aspectos Tecnicos:**
Combina `POLIZA_SEGURO` con `CONTRATO`, calculando el costo diario promedio de prima por tipo de cobertura y el valor total de contratos amparados.


````python
query_8 = """
SELECT
    ps.Aseguradora,
    ps.TipoCobertura,
    COUNT(ps.CodPoliza) AS CantidadPolizas,
    ROUND(AVG(ps.CostoDiario), 2) AS CostoDiarioPromedio,
    ROUND(SUM(c.CostoTotalReal), 2) AS TotalFacturadoAsociado
FROM POLIZA_SEGURO ps
JOIN CONTRATO c ON ps.NumContrato = c.NumContrato
GROUP BY ps.Aseguradora, ps.TipoCobertura
ORDER BY CantidadPolizas DESC, CostoDiarioPromedio DESC;
"""
df_8 = pd.read_sql_query(query_8, conn)
df_8
````

### Consulta 9: Vehiculos con Costo de Mantenimiento Superior al Promedio de la Flota
**Problema de Negocio que Responde:**
Detecta de forma proactiva vehiculos criticos cuyos gastos acumulados de taller sobrepasan la media de la compania, permitiendo programar auditorias mecanicas preventivas o evaluar garantias con fabricantes.

**Aspectos Tecnicos:**
Emplea una subconsulta escalar en la clausula `HAVING` para filtrar dinamicamente aquellos vehiculos cuya sumatoria de costos de mantenimiento supere el promedio general (`AVG(CostoTotalMant)`).


````python
query_9 = """
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    s.Ciudad AS CiudadSede,
    ROUND(SUM(m.CostoTotalMant), 2) AS GastoMantenimiento
FROM VEHICULO v
JOIN SUCURSAL s ON v.IdSucursal = s.IdSucursal
JOIN MANTENIMIENTO m ON v.VIN = m.VIN
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, s.Ciudad
HAVING SUM(m.CostoTotalMant) > (
    SELECT AVG(CostoTotalMant) FROM MANTENIMIENTO
)
ORDER BY GastoMantenimiento DESC;
"""
df_9 = pd.read_sql_query(query_9, conn)
df_9
````

### Consulta 10: Tasa de Conversion y Efectividad de Reservas Frente a Alquileres Reales
**Problema de Negocio que Responde:**
Monitorea el embudo operativo del negocio, analizando que porcentaje de reservas planificadas se concretaron efectivamente en contratos de alquiler frente a reservas pendientes o en estado de confirmacion.

**Aspectos Tecnicos:**
Realiza un `LEFT JOIN` entre `RESERVA` y `ALQUILER_EFECTIVO`, agrupando por estado y computando la tasa porcentual de conversion mediante operaciones aritmeticas con conteos condicionales.


````python
query_10 = """
SELECT
    r.Estado AS EstadoReserva,
    COUNT(r.CodReserva) AS CantidadReservas,
    COUNT(ae.IdAlquiler) AS AlquileresMaterializados,
    ROUND(COUNT(ae.IdAlquiler) * 100.0 / COUNT(r.CodReserva), 2) AS PorcentajeEfectividad
FROM RESERVA r
LEFT JOIN ALQUILER_EFECTIVO ae ON r.CodReserva = ae.CodReserva
GROUP BY r.Estado
ORDER BY CantidadReservas DESC;
"""
df_10 = pd.read_sql_query(query_10, conn)
df_10
````
