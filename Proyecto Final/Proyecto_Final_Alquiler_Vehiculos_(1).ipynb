## Descripción del Proyecto

Proyecto de Bases de Datos Relacionales orientado al diseño e implementación de una plataforma integral para la reserva, formalización de contratos, monitoreo logístico de flota y mantenimiento de diferentes tipos de vehículos en varias ciudades del país.

El sistema resuelve problemáticas críticas del sector mediante la automatización y centralización de datos: garantiza el control de disponibilidad en tiempo real para evitar la sobrerreserva (overbooking), gestiona el mantenimiento preventivo y correctivo de la flota, administra de forma segmentada a los clientes (ocasionales y empresariales) junto con sus coberturas de seguro y membresías activas, y proporciona soporte analítico mediante consultas SQL estructuradas.

# 1. Análisis de la Situación

### 1.1 Contexto del Problema y Justificación

#### Descripción del Escenario Operativo
La empresa de alquiler de vehículos se enfrenta a un escenario de crecimiento operativo acelerado que sobrepasa sus capacidades de gestión manuales o semiautomatizadas. Actualmente, el flujo operativo depende de procesos descentralizados, lo que genera inconsistencias críticas en el manejo de la información transaccional y de la flota.

#### Problemáticas Principales Identificadas

* **Inconsistencia en Reservas y Disponibilidad:** Falta de validación en tiempo real que causa sobrerreservas (*overbooking*) o la asignación de vehículos no disponibles (por estar en mantenimiento preventivo o en tránsito entre sucursales).
* **Descontrol en la Flota Diversificada:** Incapacidad de gestionar eficientemente las especificaciones únicas de cada segmento de vehículo (p. ej., tiempos e infraestructura de carga para vehículos eléctricos, límites de tonelaje para vans de carga y especificaciones técnicas para automóviles particulares).
* **Gestión Deficiente de Clientes y Beneficios:** Dificultad para segmentar adecuadamente el comportamiento de clientes ocasionales frente a empresariales, impidiendo aplicar reglas de negocio para descuentos corporativos o programas de referidos (`RECOMIENDA`).
* **Falta de Trazabilidad Contractual y Mantenimiento:** Pérdida de seguimiento a los contratos activos, seguros vinculados y registros operativos de los talleres internos de cada sucursal.

#### Objetivo del Proyecto
Diseñar e implementar un sistema de base de datos relacional modelado en tercera forma normal (3FN) y soportado por un diagrama E-R extendido. Este sistema centraliza la gestión de flotas, controla la disponibilidad en tiempo real mediante la integración de la agregación `AlquilerEfectivo` y garantiza la integridad referencial en todo el ciclo de vida del alquiler.

---

### 1.2 Tendencias Actuales del Sector

El sector del alquiler de vehículos atraviesa una transformación tecnológica orientada a la automatización y la optimización operativa. Las principales tendencias tecnológicas con impacto directo en el diseño de bases de datos son:

* **Electrificación e Integración Tecnológica de Flotas:** La inclusión de vehículos eléctricos requiere el registro y monitoreo de variables críticas como el nivel de batería, autonomía restante, infraestructura de carga y tiempos de recarga estimados (atributos integrados en la entidad `VEHICULO_ELECTRICO`).
* **Telemetría y Monitoreo IoT en Tiempo Real:** El uso de dispositivos GPS y sensores a bordo genera un flujo constante de datos sobre kilometraje, estado del motor y ubicación geográfica, lo que demanda un historial estructurado de mantenimientos en `TALLER`.
* **Modelos Flexibles de Movilidad y Suscripción:** La transición del alquiler tradicional hacia esquemas de suscripción mensual o corporativa exige esquemas de segmentación de clientes (`CLIENTE_EMPRESARIAL` vs. `CLIENTE_OCASIONAL`) y manejo de beneficios o descuentos parametrizables.
* **Precios Dinámicos y Contratación Digital:** La automatización de contratos y pólizas de seguro vinculadas a la reserva requiere estructuras relacionales de tipo N:M (como `CONTRATO` e `INCLUYE` con `POLIZA_SEGURO`) para adaptar cobros según demanda y nivel de cobertura.

### Descripción Estructural del Modelo

* **Jerarquías de Especialización (Total y Disjunta):**
  * **CLIENTE:** Se especializa disjuntamente en `CLIENTE OCASIONAL` (atributos específicos: Tarjeta de Crédito, Licencia de Conducción) y `CLIENTE EMPRESARIAL` (Razón Social, Descuento Corporativo, Contacto Ejecutivo).
  * **VEHICULO:** Se especializa en `AUTOMOVIL` (Número de Pasajeros, Tipo de Carrocería), `VAN_CARGA` (Capacidad de Carga en Kg, Volumen Útil en M3, Categoría de Licencia requerida), `VEHICULO_ELECTRICO` (Autonomía en Km, Tiempo de Carga en Horas) y `VEHICULO_PREMIUM` (Torque, aceleración y detalles).

* **Entidad Débil:**
  * `TALLER` se modeló inicialmente como entidad débil dependiente de la entidad fuerte `SUCURSAL` mediante la relación `POSEE` (1:N), utilizando `Cod_Taller` como discriminador parcial (concepto que en la fase relacional se evolucionó hacia talleres independientes para dar soporte a redes externas).

* **Mecanismo de Agregación (Bloque Operativo Central):**
  * La transacción central del sistema consolida una relación ternaria `AlquilerEfectivo` entre `CLIENTE` (1:N), `VEHICULO` (1:N) y `RESERVA` (1:1).
  * Dicho grupo está delimitado dentro del **Marco de Agregación**, del cual deriva la relación 1:1 `FORMALIZA EN` conectada a la entidad `CONTRATO`. Esto garantiza que ningún contrato se genere sin la presencia física validada de la reserva, el cliente y el vehículo entregado.

* **Relaciones Adicionales y Reglas de Negocio:**
  * **Relación Reflexiva:** `RECOMIENDA` sobre `CLIENTE` con roles definidos (0:N Recomendador / 0:1 Recomendado), modelando programas de referidos comerciales de forma limpia e independiente.
  * **Cobertura de Seguros:** La entidad `CONTRATO` se vincula con `POLIZA_SEGURO` para especificar la aseguradora y el valor diario del amparo.

# 3. Modelo Relacional y su Normalización

### 3.1 Paso de Modelo E-R a Modelo Relacional Inicial

Al realizar la transformación del Modelo Entidad-Relación Extendido al Modelo Relacional, se aplicaron mejoras técnicas orientadas a la viabilidad operativa en entornos reales:

1. **Clave Primaria de Vehículos:** Se estableció el `VIN` (*Vehicle Identification Number*) como Clave Primaria (PK) en la superclase `VEHICULO`. La placa es un identificador de tránsito susceptible a cambios administrativos o reasignaciones, mientras que el VIN es inmutable legal y físicamente.
2. **Aislamiento de Atributos Especializados:** Para evitar valores nulos masivos en vehículos de combustión tradicional, los atributos exclusivos de carga eléctrica se aislaron en la subclase `VEHICULO_ELECTRICO`.
3. **Clave Subrogada en Alquiler:** La agregación ternaria central `AlquilerEfectivo` une Cliente, Reserva y Vehículo. En lugar de arrastrar una clave compuesta masiva, se definió la clave artificial `IdAlquiler`, optimizando la conexión con `CONTRATO`.
4. **Logística de Talleres y Mantenimiento:** Se desacopló el `TALLER` de la `SUCURSAL` para permitir el trabajo con talleres externos aliados o concesionarios. Además, el mantenimiento se separó en cabecera (`MANTENIMIENTO`) y detalle (`DETALLE_MANTENIMIENTO`) para discriminar consumos de repuestos y mano de obra.
5. **Control de Reservas:** Se conectó la tabla `RESERVA` directamente con el `CLIENTE` mediante `NIT_Cliente`, asegurando la trazabilidad de la intencion de alquiler antes de que se materialice la entrega del automotor.

### 3.2 Proceso de Normalización (1FN, 2FN, 3FN)

<p align="center">
<img src='https://github.com/Strayker54/Proyecto-Alquiler-de-Vehiculos-BD-relacionales/blob/main/Modelo%20Relacional/Modelo%20Relacional%20Normalizado%20new.png?raw=true' alt='Modelo Relacional Normalizado' width='1000'>
</p>

#### Paso 1: Primera Forma Normal (1FN)
* **Regla Técnica:** Una tabla está en 1FN si y solo si todos sus atributos son atómicos; es decir, no contiene campos compuestos ni grupos repetitivos o multivaluados.
* **Diagnóstico en el Modelo Inicial:** La entidad `CLIENTE` violaba la 1FN en dos campos:
  1. `Direccion_Compuesta`: Almacenaba simultáneamente ciudad, barrio y calle en una sola cadena. Se solucionó dividiendo el campo en tres columnas atómicas: `Ciudad`, `Barrio` y `Calle`.
  2. `Telefonos`: Almacenaba una colección variable de teléfonos para un mismo cliente, rompiendo la atomicidad. Se resolvió creando la tabla dependiente `CLIENTE_TELEFONO` con clave primaria compuesta `(NIT, Telefono)`.

#### Paso 2: Segunda Forma Normal (2FN)
* **Regla Técnica:** Una relación está en 2FN si cumple 1FN y todos los atributos no clave tienen dependencia funcional completa respecto a la clave primaria. Se analiza específicamente en relaciones con **claves compuestas** para descartar dependencias parciales.
* **Diagnóstico en el Modelo:**
  1. `CONTRATA`: Clave primaria `(NIT, NombreUnico)`. Los atributos `FechaInicio` y `FechaFin` representan la vigencia de esa membresía particular tomada por ese cliente específico. Dependen de ambos componentes de la clave.
  2. `CLIENTE_TELEFONO`: Clave primaria `(NIT, Telefono)`. Al no poseer atributos descriptivos fuera de la clave, cumple 2FN de forma directa.
  3. Todas las demás tablas poseen claves primarias simples de un solo atributo, por lo que satisfacen la 2FN por definición.

#### Paso 3: Tercera Forma Normal (3FN)
* **Regla Técnica:** Una relación está en 3FN si cumple 2FN y no contiene dependencias transitivas; ningún atributo no clave puede depender funcionalmente de otro atributo no clave.
* **Diagnóstico y Soluciones Aplicadas:**
  1. **Independencia de Talleres:** Si se almacenaran datos como `CiudadTaller` o `DireccionTaller` en `MANTENIMIENTO`, dependerían de `CodTaller` y no directamente de `IdMantenimiento`. La existencia de la tabla independiente `TALLER` asegura que los datos de localización dependan únicamente de su clave primaria.
  2. **Especialización de Flota:** Incluir especificaciones como `AutonomiaKm` o `VolumenUtilM3` en `VEHICULO` crearía inconsistencias y celdas con valores nulos para modelos a gasolina o automóviles estándar. Las tablas hijas `AUTOMOVIL`, `VAN_CARGA`, `VEHICULO_ELECTRICO` y `VEHICULO_PREMIUM` garantizan que cada atributo especializado dependa de forma directa y no transitiva de su `VIN`.
  3. **Agregación de Alquiler:** El uso de `ALQUILER_EFECTIVO` con clave subrogada `IdAlquiler` desacopla al `CONTRATO` y las `POLIZA_SEGURO` de los detalles internos del cliente y del vehículo, eliminando transitividades en la facturación.

# 4. Base de Datos Construida en SQL (Esquema DDL)

Para la implementación física del modelo relacional se seleccionó el motor **SQLite**, un sistema de gestión de bases de datos relacionales embebido y robusto que cumple con los estándares SQL ANSI.

Un aspecto técnico fundamental en SQLite es que el soporte para claves foráneas viene desactivado por defecto para mantener retrocompatibilidad histórica. Por esta razón, en la primera celda de ejecución se activa explícitamente la directiva:
```sql
PRAGMA foreign_keys = ON;
```
Esto garantiza que cualquier intento de violar la integridad referencial (por ejemplo, insertar un vehículo con una sucursal inexistente o asociar un contrato a un alquiler inválido) sea bloqueado inmediatamente por el motor.

A continuación, se ejecuta la creación de la base de datos `alquiler_vehiculos.db` y las sentencias DDL correspondientes a las 19 tablas normalizadas.


````python
import sqlite3
import pandas as pd
import os

db_path = "alquiler_vehiculos.db"
conn = sqlite3.connect(db_path)

conn.execute("PRAGMA foreign_keys = ON;")
cursor = conn.cursor()

ddl_script = """
CREATE TABLE IF NOT EXISTS SUCURSAL (
    IdSucursal INTEGER PRIMARY KEY,
    Ciudad VARCHAR(50) NOT NULL,
    Direccion VARCHAR(150) NOT NULL
);

CREATE TABLE IF NOT EXISTS TALLER (
    CodTaller INTEGER PRIMARY KEY,
    NIT_Taller INTEGER,
    NombreTaller VARCHAR(100) NOT NULL,
    TipoTaller VARCHAR(20) NOT NULL,
    Ciudad VARCHAR(50) NOT NULL,
    Direccion VARCHAR(150) NOT NULL
);

CREATE TABLE IF NOT EXISTS MEMBRESIA (
    NombreUnico VARCHAR(50) PRIMARY KEY,
    CostoMensual DECIMAL(10,2) NOT NULL,
    Beneficios TEXT NOT NULL
);

CREATE TABLE IF NOT EXISTS CLIENTE (
    NIT INTEGER PRIMARY KEY,
    CUC VARCHAR(20) UNIQUE NOT NULL,
    Nombre VARCHAR(100) NOT NULL,
    Ciudad VARCHAR(50) NOT NULL,
    Barrio VARCHAR(50) NOT NULL,
    Calle VARCHAR(100) NOT NULL,
    Recomendador_NIT INTEGER,
    FOREIGN KEY (Recomendador_NIT) REFERENCES CLIENTE(NIT)
);

CREATE TABLE IF NOT EXISTS CLIENTE_TELEFONO (
    NIT INTEGER NOT NULL,
    Telefono VARCHAR(20) NOT NULL,
    PRIMARY KEY (NIT, Telefono),
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS CLIENTE_OCASIONAL (
    NIT INTEGER PRIMARY KEY,
    TarjetaCredito VARCHAR(20) NOT NULL,
    LicenciaConduccion VARCHAR(20) NOT NULL,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS CLIENTE_EMPRESARIAL (
    NIT INTEGER PRIMARY KEY,
    RazonSocial VARCHAR(150) NOT NULL,
    DescuentoCorporativo DECIMAL(5,2) NOT NULL,
    ContactoEjecutivo VARCHAR(100) NOT NULL,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS CONTRATA (
    NIT INTEGER NOT NULL,
    NombreUnico VARCHAR(50) NOT NULL,
    FechaInicio DATE NOT NULL,
    FechaFin DATE NOT NULL,
    PRIMARY KEY (NIT, NombreUnico),
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (NombreUnico) REFERENCES MEMBRESIA(NombreUnico)
);

CREATE TABLE IF NOT EXISTS VEHICULO (
    VIN VARCHAR(17) PRIMARY KEY,
    Placa VARCHAR(10) UNIQUE NOT NULL,
    Marca VARCHAR(50) NOT NULL,
    Modelo VARCHAR(50) NOT NULL,
    Año INTEGER NOT NULL,
    KilometrajeActual INTEGER NOT NULL,
    TarifaDiariaBase DECIMAL(10,2) NOT NULL,
    NumPasajeros INTEGER NOT NULL,
    TipoTransmision VARCHAR(20) NOT NULL,
    TipoTraccion VARCHAR(10) NOT NULL,
    TipoCombustible VARCHAR(20) NOT NULL,
    IdSucursal INTEGER NOT NULL,
    FOREIGN KEY (IdSucursal) REFERENCES SUCURSAL(IdSucursal)
);

CREATE TABLE IF NOT EXISTS AUTOMOVIL (
    VIN VARCHAR(17) PRIMARY KEY,
    TipoCarroceria VARCHAR(30) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS VAN_CARGA (
    VIN VARCHAR(17) PRIMARY KEY,
    CapacidadCargaKg DECIMAL(8,2) NOT NULL,
    VolumenUtilM3 DECIMAL(6,2) NOT NULL,
    TipoLicencia VARCHAR(5) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS VEHICULO_ELECTRICO (
    VIN VARCHAR(17) PRIMARY KEY,
    AutonomiaKm INTEGER NOT NULL,
    TiempoCargaHoras DECIMAL(4,2) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS VEHICULO_PREMIUM (
    VIN VARCHAR(17) PRIMARY KEY,
    Torque VARCHAR(30) NOT NULL,
    Aceleracion VARCHAR(30) NOT NULL,
    DetallesLujo TEXT NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN) ON DELETE CASCADE
);

CREATE TABLE IF NOT EXISTS RESERVA (
    CodReserva INTEGER PRIMARY KEY,
    NIT_Cliente INTEGER NOT NULL,
    VIN_Solicitado VARCHAR(17),
    FechaCreacion DATE NOT NULL,
    FechaInicio DATE NOT NULL,
    FechaFin DATE NOT NULL,
    Estado VARCHAR(20) NOT NULL,
    FOREIGN KEY (NIT_Cliente) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (VIN_Solicitado) REFERENCES VEHICULO(VIN)
);

CREATE TABLE IF NOT EXISTS ALQUILER_EFECTIVO (
    IdAlquiler INTEGER PRIMARY KEY AUTOINCREMENT,
    NIT INTEGER NOT NULL,
    CodReserva INTEGER NOT NULL,
    VIN VARCHAR(17) NOT NULL,
    FechaHoraSalida DATETIME NOT NULL,
    NivelCombSalida VARCHAR(20) NOT NULL,
    KmSalida INTEGER NOT NULL,
    FechaHoraLlegada DATETIME,
    NivelCombLlegada VARCHAR(20),
    KmLlegada INTEGER,
    NotasEstado TEXT,
    FOREIGN KEY (NIT) REFERENCES CLIENTE(NIT),
    FOREIGN KEY (CodReserva) REFERENCES RESERVA(CodReserva),
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN)
);

CREATE TABLE IF NOT EXISTS CONTRATO (
    NumContrato INTEGER PRIMARY KEY,
    IdAlquiler INTEGER NOT NULL,
    FechaFirma DATE NOT NULL,
    Condiciones TEXT NOT NULL,
    CostoTotalReal DECIMAL(12,2) NOT NULL,
    FOREIGN KEY (IdAlquiler) REFERENCES ALQUILER_EFECTIVO(IdAlquiler)
);

CREATE TABLE IF NOT EXISTS POLIZA_SEGURO (
    CodPoliza INTEGER PRIMARY KEY,
    NumContrato INTEGER NOT NULL,
    Aseguradora VARCHAR(100) NOT NULL,
    TipoCobertura VARCHAR(50) NOT NULL,
    CostoDiario DECIMAL(10,2) NOT NULL,
    FOREIGN KEY (NumContrato) REFERENCES CONTRATO(NumContrato)
);

CREATE TABLE IF NOT EXISTS MANTENIMIENTO (
    IdMantenimiento INTEGER PRIMARY KEY,
    VIN VARCHAR(17) NOT NULL,
    CodTaller INTEGER NOT NULL,
    FechaIngreso DATE NOT NULL,
    CostoTotalMant DECIMAL(12,2) NOT NULL,
    FOREIGN KEY (VIN) REFERENCES VEHICULO(VIN),
    FOREIGN KEY (CodTaller) REFERENCES TALLER(CodTaller)
);

CREATE TABLE IF NOT EXISTS DETALLE_MANTENIMIENTO (
    IdDetalle INTEGER PRIMARY KEY,
    IdMantenimiento INTEGER NOT NULL,
    RepuestoOServicio VARCHAR(100) NOT NULL,
    Cantidad INTEGER NOT NULL,
    CostoUnitario DECIMAL(10,2) NOT NULL,
    Subtotal DECIMAL(10,2) NOT NULL,
    FOREIGN KEY (IdMantenimiento) REFERENCES MANTENIMIENTO(IdMantenimiento)
);
"""

cursor.executescript(ddl_script)
conn.commit()

tablas = cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';").fetchall()
print(f"Esquema DDL ejecutado correctamente. Total tablas en la base de datos: {len(tablas)}")
for t in sorted([item[0] for item in tablas]):
    print(f"  - {t}")
````

# 6. Consultas Estratégicas sobre la Base de Datos (10 Consultas SQL)

Para explotar analíticamente el modelo implementado y responder a preguntas de decisión gerencial y logística, se diseñaron **10 consultas SQL avanzadas**. Las consultas utilizan técnicas del estándar relacional: combinaciones internas y externas (`INNER JOIN`, `LEFT JOIN`), agrupaciones con funciones de agregación (`COUNT`, `SUM`, `AVG`, `ROUND`), filtros sobre agregaciones (`HAVING`), funciones condicionales (`COALESCE`), concatenaciones de grupos (`GROUP_CONCAT`) y subconsultas escalares.

Cada consulta es ejecutada mediante `pandas.read_sql_query`, presentando la información de forma clara y tabular.


### Consulta 1: Ingresos Totales y Ticket Promedio por Sucursal Comercial
**Problema de Negocio que Responde:**
Permite a la dirección financiera evaluar el desempeño comercial y la rentabilidad por ciudad, identificando cuáles sedes están generando mayor volumen de facturación neta y cuál es el valor promedio percibido por alquiler (*ticket promedio*).

**Aspectos Técnicos:**
Combina cuatro tablas (`SUCURSAL`, `VEHICULO`, `ALQUILER_EFECTIVO`, `CONTRATO`) mediante `JOIN`, agrupando por sede y calculando agregaciones monetarias con redondeo a dos decimales.


````python
query_1 = """
SELECT
    s.IdSucursal,
    s.Ciudad,
    COUNT(c.NumContrato) AS TotalAlquileres,
    ROUND(SUM(c.CostoTotalReal), 2) AS IngresosTotales,
    ROUND(AVG(c.CostoTotalReal), 2) AS TicketPromedio
FROM SUCURSAL s
JOIN VEHICULO v ON s.IdSucursal = v.IdSucursal
JOIN ALQUILER_EFECTIVO ae ON v.VIN = ae.VIN
JOIN CONTRATO c ON ae.IdAlquiler = c.IdAlquiler
GROUP BY s.IdSucursal, s.Ciudad
ORDER BY IngresosTotales DESC;
"""
df_1 = pd.read_sql_query(query_1, conn)
df_1
````

### Consulta 2: Historial y Costos de Mantenimiento y Repuestos por Vehículo
**Problema de Negocio que Responde:**
Permite al departamento de taller y flota monitorear el gasto acumulado en reparaciones y repuestos para cada automotor, identificando qué taller realizó la labor y cuántos conceptos técnicos fueron intervenidos.

**Aspectos Técnicos:**
Realiza una unión de cuatro tablas (`VEHICULO`, `MANTENIMIENTO`, `TALLER`, `DETALLE_MANTENIMIENTO`), agrupando por unidad vehicular y calculando el costo total a partir de la suma de subtotales de repuestos.


````python
query_2 = """
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    t.NombreTaller,
    m.FechaIngreso,
    COUNT(dm.IdDetalle) AS TotalItemsAtendidos,
    ROUND(SUM(dm.Subtotal), 2) AS CostoTotalReparaciones
FROM VEHICULO v
JOIN MANTENIMIENTO m ON v.VIN = m.VIN
JOIN TALLER t ON m.CodTaller = t.CodTaller
JOIN DETALLE_MANTENIMIENTO dm ON m.IdMantenimiento = dm.IdMantenimiento
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, t.NombreTaller, m.FechaIngreso
ORDER BY CostoTotalReparaciones DESC;
"""
df_2 = pd.read_sql_query(query_2, conn)
df_2
````

### Consulta 3: Clientes Empresariales con Membresías Vigentes y Descuento Corporativo
**Problema de Negocio que Responde:**
Permite al área comercial y de fidelización verificar qué clientes corporativos mantienen membresías activas, qué plan tienen contratado y qué porcentaje de descuento corresponde aplicar sobre sus cotizaciones de flota.

**Aspectos Técnicos:**
Une la subclase `CLIENTE_EMPRESARIAL` con la tabla relacional `CONTRATA` y el catálogo `MEMBRESIA`, filtrando suscripciones vigentes mediante comparación de fechas.


````python
query_3 = """
SELECT
    ce.NIT,
    ce.RazonSocial,
    ce.DescuentoCorporativo AS PorcentajeDescuento,
    m.NombreUnico AS PlanMembresia,
    m.CostoMensual,
    co.FechaInicio AS VigenciaDesde,
    co.FechaFin AS VigenciaHasta
FROM CLIENTE_EMPRESARIAL ce
JOIN CONTRATA co ON ce.NIT = co.NIT
JOIN MEMBRESIA m ON co.NombreUnico = m.NombreUnico
WHERE DATE('now') BETWEEN co.FechaInicio AND co.FechaFin
   OR co.FechaFin >= '2026-11-01'
ORDER BY ce.DescuentoCorporativo DESC;
"""
df_3 = pd.read_sql_query(query_3, conn)
df_3
````

Aplica `LEFT JOIN` partiendo desde `VEHICULO` hacia `ALQUILER_EFECTIVO` y `CONTRATO` para incluir unidades sin alquileres, utilizando `COALESCE` para manejar nulos contables y ordenando por frecuencia de despacho.

### Consulta 7: Red de Fidelización y Clientes Referidos Mediante Relación Reflexiva

### Consulta 8: Cobertura de Pólizas de Seguro por Aseguradora y Costo Promedio Diario

### Consulta 9: Vehículos con Costo de Mantenimiento Superior al Promedio de la Flota
**Problema de Negocio que Responde:**
Detecta de forma proactiva vehículos críticos cuyos gastos acumulados de taller sobrepasan la media de la compañía, permitiendo programar auditorías mecánicas preventivas o evaluar garantías con fabricantes.

**Aspectos Técnicos:**
Emplea una subconsulta escalar en la cláusula `HAVING` para filtrar dinámicamente aquellos vehículos cuya sumatoria de costos de mantenimiento supere el promedio general (`AVG(CostoTotalMant)`).


````python
query_9 = """
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    s.Ciudad AS CiudadSede,
    ROUND(SUM(m.CostoTotalMant), 2) AS GastoMantenimiento
FROM VEHICULO v
JOIN SUCURSAL s ON v.IdSucursal = s.IdSucursal
JOIN MANTENIMIENTO m ON v.VIN = m.VIN
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, s.Ciudad
HAVING SUM(m.CostoTotalMant) > (
    SELECT AVG(CostoTotalMant) FROM MANTENIMIENTO
)
ORDER BY GastoMantenimiento DESC;
"""
df_9 = pd.read_sql_query(query_9, conn)
df_9
````

### Consulta 10: Tasa de Conversión y Efectividad de Reservas Frente a Alquileres Reales
**Problema de Negocio que Responde:**
Monitorea el embudo operativo del negocio, analizando qué porcentaje de reservas planificadas se concretaron efectivamente en contratos de alquiler frente a reservas pendientes o en estado de confirmación.

**Aspectos Técnicos:**
Realiza un `LEFT JOIN` entre `RESERVA` y `ALQUILER_EFECTIVO`, agrupando por estado y computando la tasa porcentual de conversión mediante operaciones aritméticas con conteos condicionales.


````python
query_10 = """
SELECT
    r.Estado AS EstadoReserva,
    COUNT(r.CodReserva) AS CantidadReservas,
    COUNT(ae.IdAlquiler) AS AlquileresMaterializados,
    ROUND(COUNT(ae.IdAlquiler) * 100.0 / COUNT(r.CodReserva), 2) AS PorcentajeEfectividad
FROM RESERVA r
LEFT JOIN ALQUILER_EFECTIVO ae ON r.CodReserva = ae.CodReserva
GROUP BY r.Estado
ORDER BY CantidadReservas DESC;
"""
df_10 = pd.read_sql_query(query_10, conn)
df_10
````

````python
# =============================================================================
# MÓDULO INTERACTIVO DE CONSULTAS - PLATAFORMA DE ALQUILER DE VEHÍCULOS
# Google Colab | SQLite + Pandas + ipywidgets
# Editor SQL mostrado directamente para evitar problemas de foco en Colab.
#
# Inserta esta celda al FINAL del cuaderno, después del DDL, DML y las 10 consultas.
# El cuaderno puede haber ejecutado conn.close(): este módulo abre su propia
# conexión de solo lectura usando db_path y nunca reutiliza ni cierra conn.
# Este módulo no crea ni modifica tablas ni registros.
# =============================================================================

import os
import re
import html
import sqlite3
from pathlib import Path
from datetime import datetime

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

try:
    from google.colab import files as colab_files
    from google.colab import output as colab_output
    # Activa el administrador de widgets en Google Colab.
    colab_output.enable_custom_widget_manager()
except ImportError:
    colab_files = None


# -----------------------------------------------------------------------------
# 1. CONFIGURACIÓN Y CONEXIÓN DE SOLO LECTURA
# -----------------------------------------------------------------------------

DEFAULT_DB_PATH = "alquiler_vehiculos.db"
# Reutiliza la variable db_path definida en las celdas anteriores, si existe.
DB_PATH = str(globals().get("db_path", DEFAULT_DB_PATH) or DEFAULT_DB_PATH)
MAX_RESULT_ROWS = 10000

EXPECTED_TABLES = [
    "ALQUILER_EFECTIVO",
    "AUTOMOVIL",
    "CLIENTE",
    "CLIENTE_EMPRESARIAL",
    "CLIENTE_OCASIONAL",
    "CLIENTE_TELEFONO",
    "CONTRATA",
    "CONTRATO",
    "DETALLE_MANTENIMIENTO",
    "MANTENIMIENTO",
    "MEMBRESIA",
    "POLIZA_SEGURO",
    "RESERVA",
    "SUCURSAL",
    "TALLER",
    "VAN_CARGA",
    "VEHICULO",
    "VEHICULO_ELECTRICO",
    "VEHICULO_PREMIUM",
]


def resolve_db_path():
    """Resuelve la ruta absoluta sin crear accidentalmente una BD vacía."""
    return Path(DB_PATH).expanduser().resolve()


def get_read_only_connection(use_authorizer=False):
    """Abre una conexión SQLite en modo de solo lectura.

    La conexión nunca sustituye ni cierra la variable conn creada en el
    cuaderno original. El autorizador se activa para consultas del usuario.
    """
    db_file = resolve_db_path()
    if not db_file.is_file():
        raise FileNotFoundError(
            f"No se encontró la base de datos en: {db_file}. "
            "Ejecuta primero las celdas de creación y carga de datos."
        )

    uri = db_file.as_uri() + "?mode=ro"
    connection = sqlite3.connect(uri, uri=True, timeout=10)
    connection.row_factory = sqlite3.Row
    connection.execute("PRAGMA query_only = ON;")

    if use_authorizer:
        connection.set_authorizer(read_only_authorizer)

    return connection


def read_only_authorizer(action, arg1, arg2, database_name, trigger_name):
    """Autoriza únicamente operaciones de lectura durante SQL personalizado.

    READ, SELECT, FUNCTION y RECURSIVE son necesarias para SELECT, JOIN,
    agregaciones y CTE. Cualquier operación no incluida se rechaza.
    """
    allowed_actions = {
        sqlite3.SQLITE_SELECT,
        sqlite3.SQLITE_READ,
        sqlite3.SQLITE_FUNCTION,
    }

    recursive_action = getattr(sqlite3, "SQLITE_RECURSIVE", None)
    if recursive_action is not None:
        allowed_actions.add(recursive_action)

    if action == sqlite3.SQLITE_FUNCTION:
        function_name = str(arg2 or arg1 or "").lower()
        if function_name in {"load_extension", "writefile", "readfile"}:
            return sqlite3.SQLITE_DENY

    return sqlite3.SQLITE_OK if action in allowed_actions else sqlite3.SQLITE_DENY


def verificar_estado_sistema():
    """Comprueba archivo, integridad básica y presencia de las 19 tablas."""
    db_file = resolve_db_path()
    if not db_file.is_file():
        return (
            False,
            f"No existe el archivo de base de datos: {db_file}. "
            "Ejecuta primero las celdas de construcción y carga.",
        )

    connection = None
    try:
        connection = get_read_only_connection()
        integrity = connection.execute("PRAGMA quick_check;").fetchone()[0]
        tables = connection.execute(
            "SELECT name FROM sqlite_master "
            "WHERE type = 'table' AND name NOT LIKE 'sqlite_%';"
        ).fetchall()
        table_names = {row[0].upper() for row in tables}
        missing = [name for name in EXPECTED_TABLES if name not in table_names]

        fk_violations = connection.execute("PRAGMA foreign_key_check;").fetchall()

        if integrity != "ok":
            return False, f"La comprobación de integridad de SQLite reportó: {integrity}"
        if missing:
            return False, "Faltan tablas del esquema: " + ", ".join(missing)
        if fk_violations:
            return (
                False,
                "Se detectaron inconsistencias de integridad referencial "
                f"({len(fk_violations)} registros). Revisa las claves foráneas.",
            )

        return (
            True,
            f"Conexión de solo lectura correcta. Se detectaron las "
            f"{len(EXPECTED_TABLES)} tablas esperadas y no hay violaciones "
            f"de claves foráneas. Archivo: {db_file.name}",
        )
    except Exception as exc:
        return False, f"No fue posible verificar la base de datos: {exc}"
    finally:
        if connection is not None:
            connection.close()


# -----------------------------------------------------------------------------
# 2. LAS 10 CONSULTAS PREDEFINIDAS
#    Se conserva su propósito y se corrigen las condiciones con errores lógicos.
# -----------------------------------------------------------------------------

PREDEFINED_QUERIES = {
    "1. Ingresos y ticket promedio por sucursal base": {
        "desc": (
            "Resume los contratos asociados a vehículos según la sucursal "
            "base registrada para cada vehículo. No representa necesariamente "
            "la sucursal física de recogida, porque el modelo no almacena ese dato."
        ),
        "sql": """WITH contratos_por_alquiler AS (
    SELECT IdAlquiler, SUM(CostoTotalReal) AS TotalContrato
    FROM CONTRATO
    GROUP BY IdAlquiler
)
SELECT
    s.IdSucursal,
    s.Ciudad,
    COUNT(DISTINCT ae.IdAlquiler) AS TotalAlquileres,
    ROUND(SUM(cpa.TotalContrato), 2) AS IngresosTotales,
    ROUND(AVG(cpa.TotalContrato), 2) AS TicketPromedio
FROM SUCURSAL AS s
JOIN VEHICULO AS v ON v.IdSucursal = s.IdSucursal
JOIN ALQUILER_EFECTIVO AS ae ON ae.VIN = v.VIN
JOIN contratos_por_alquiler AS cpa ON cpa.IdAlquiler = ae.IdAlquiler
GROUP BY s.IdSucursal, s.Ciudad
ORDER BY IngresosTotales DESC;""",
    },
    "2. Historial y costos de mantenimiento": {
        "desc": (
            "Muestra cada orden de mantenimiento, el vehículo atendido, el taller, "
            "la cantidad de conceptos registrados y el costo calculado a partir "
            "de los subtotales de sus detalles."
        ),
        "sql": """SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    m.IdMantenimiento,
    t.NombreTaller,
    m.FechaIngreso,
    COUNT(dm.IdDetalle) AS TotalItemsAtendidos,
    ROUND(COALESCE(SUM(dm.Subtotal), 0), 2) AS CostoTotalReparaciones
FROM VEHICULO AS v
JOIN MANTENIMIENTO AS m ON m.VIN = v.VIN
JOIN TALLER AS t ON t.CodTaller = m.CodTaller
LEFT JOIN DETALLE_MANTENIMIENTO AS dm
    ON dm.IdMantenimiento = m.IdMantenimiento
GROUP BY
    v.VIN, v.Placa, v.Marca, v.Modelo,
    m.IdMantenimiento, t.NombreTaller, m.FechaIngreso
ORDER BY CostoTotalReparaciones DESC;""",
    },
    "3. Clientes empresariales con membresías vigentes": {
        "desc": (
            "Muestra los clientes empresariales cuyas membresías están vigentes "
            "en la fecha actual. La vigencia se determina exclusivamente con "
            "FechaInicio y FechaFin, sin fechas fijas adicionales."
        ),
        "sql": """SELECT
    ce.NIT,
    ce.RazonSocial,
    ce.DescuentoCorporativo AS PorcentajeDescuento,
    m.NombreUnico AS PlanMembresia,
    m.CostoMensual,
    co.FechaInicio AS VigenciaDesde,
    co.FechaFin AS VigenciaHasta
FROM CLIENTE_EMPRESARIAL AS ce
JOIN CONTRATA AS co ON co.NIT = ce.NIT
JOIN MEMBRESIA AS m ON m.NombreUnico = co.NombreUnico
WHERE DATE('now', '-5 hours') BETWEEN DATE(co.FechaInicio) AND DATE(co.FechaFin)
ORDER BY ce.DescuentoCorporativo DESC;""",
    },
    "4. Ficha técnica y autonomía de vehículos eléctricos": {
        "desc": (
            "Consulta la autonomía, el tiempo estimado de carga, la sucursal "
            "base y la tarifa de los vehículos eléctricos. No determina su "
            "disponibilidad temporal para una fecha de reserva."
        ),
        "sql": """SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    ve.AutonomiaKm,
    ve.TiempoCargaHoras,
    s.Ciudad AS CiudadSede,
    v.TarifaDiariaBase
FROM VEHICULO AS v
JOIN VEHICULO_ELECTRICO AS ve ON ve.VIN = v.VIN
JOIN SUCURSAL AS s ON s.IdSucursal = v.IdSucursal
ORDER BY ve.AutonomiaKm DESC;""",
    },
    "5. Kilometraje recorrido en alquileres devueltos": {
        "desc": (
            "Calcula la diferencia entre los odómetros de salida y llegada. "
            "Incluye alquileres que registran devolución y cuyo kilometraje "
            "final no es inferior al inicial."
        ),
        "sql": """SELECT
    ae.IdAlquiler,
    cl.Nombre AS Cliente,
    v.Placa,
    v.Marca || ' ' || v.Modelo AS Vehiculo,
    ae.KmSalida,
    ae.KmLlegada,
    (ae.KmLlegada - ae.KmSalida) AS KmRecorridos,
    ae.FechaHoraSalida,
    ae.FechaHoraLlegada
FROM ALQUILER_EFECTIVO AS ae
JOIN CLIENTE AS cl ON cl.NIT = ae.NIT
JOIN VEHICULO AS v ON v.VIN = ae.VIN
WHERE ae.KmLlegada IS NOT NULL
  AND ae.FechaHoraLlegada IS NOT NULL
  AND ae.KmLlegada >= ae.KmSalida
ORDER BY KmRecorridos DESC;""",
    },
    "6. Utilización y facturación por vehículo": {
        "desc": (
            "Incluye todos los vehículos, incluso los que no registran alquileres. "
            "Cuenta alquileres distintos y suma los contratos asociados sin "
            "multiplicar valores por uniones de varias filas."
        ),
        "sql": """WITH contratos_por_alquiler AS (
    SELECT IdAlquiler, SUM(CostoTotalReal) AS TotalFacturado
    FROM CONTRATO
    GROUP BY IdAlquiler
)
SELECT
    v.VIN,
    v.Placa,
    v.Marca,
    v.Modelo,
    v.TipoCombustible,
    COUNT(DISTINCT ae.IdAlquiler) AS TotalVecesAlquilado,
    ROUND(COALESCE(SUM(cpa.TotalFacturado), 0), 2) AS TotalFacturadoVehiculo
FROM VEHICULO AS v
LEFT JOIN ALQUILER_EFECTIVO AS ae ON ae.VIN = v.VIN
LEFT JOIN contratos_por_alquiler AS cpa ON cpa.IdAlquiler = ae.IdAlquiler
GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, v.TipoCombustible
ORDER BY TotalVecesAlquilado DESC, TotalFacturadoVehiculo DESC;""",
    },
    "7. Red de fidelización y clientes referidos": {
        "desc": (
            "Utiliza la relación reflexiva de CLIENTE para identificar "
            "recomendadores, contar sus referidos y concatenar sus nombres."
        ),
        "sql": """SELECT
    p.NIT AS NIT_Recomendador,
    p.Nombre AS NombreRecomendador,
    COUNT(h.NIT) AS TotalClientesReferidos,
    GROUP_CONCAT(h.Nombre, ', ') AS ClientesRecomendados
FROM CLIENTE AS p
JOIN CLIENTE AS h ON h.Recomendador_NIT = p.NIT
GROUP BY p.NIT, p.Nombre
ORDER BY TotalClientesReferidos DESC;""",
    },
    "8. Cobertura y costos de pólizas de seguro": {
        "desc": (
            "Agrupa pólizas por aseguradora y tipo de cobertura. Primero consolida "
            "las pólizas por contrato para evitar repetir el valor total de un "
            "contrato cuando tiene varias pólizas del mismo grupo."
        ),
        "sql": """WITH polizas_por_contrato AS (
    SELECT
        Aseguradora,
        TipoCobertura,
        NumContrato,
        COUNT(*) AS CantidadPolizas,
        SUM(CostoDiario) AS SumaCostoDiario
    FROM POLIZA_SEGURO
    GROUP BY Aseguradora, TipoCobertura, NumContrato
)
SELECT
    ppc.Aseguradora,
    ppc.TipoCobertura,
    SUM(ppc.CantidadPolizas) AS CantidadPolizas,
    ROUND(
        SUM(ppc.SumaCostoDiario) / NULLIF(SUM(ppc.CantidadPolizas), 0),
        2
    ) AS CostoDiarioPromedio,
    ROUND(SUM(c.CostoTotalReal), 2) AS TotalFacturadoAsociado
FROM polizas_por_contrato AS ppc
JOIN CONTRATO AS c ON c.NumContrato = ppc.NumContrato
GROUP BY ppc.Aseguradora, ppc.TipoCobertura
ORDER BY CantidadPolizas DESC, CostoDiarioPromedio DESC;""",
    },
    "9. Vehículos con gasto de mantenimiento superior al promedio": {
        "desc": (
            "Calcula primero el gasto acumulado por vehículo y luego lo compara "
            "con el promedio de los gastos acumulados de todos los vehículos "
            "que tienen registros de mantenimiento."
        ),
        "sql": """WITH gasto_por_vehiculo AS (
    SELECT
        v.VIN,
        v.Placa,
        v.Marca,
        v.Modelo,
        s.Ciudad AS CiudadSede,
        SUM(m.CostoTotalMant) AS GastoMantenimiento
    FROM VEHICULO AS v
    JOIN SUCURSAL AS s ON s.IdSucursal = v.IdSucursal
    JOIN MANTENIMIENTO AS m ON m.VIN = v.VIN
    GROUP BY v.VIN, v.Placa, v.Marca, v.Modelo, s.Ciudad
)
SELECT
    VIN,
    Placa,
    Marca,
    Modelo,
    CiudadSede,
    ROUND(GastoMantenimiento, 2) AS GastoMantenimiento
FROM gasto_por_vehiculo
WHERE GastoMantenimiento > (
    SELECT AVG(GastoMantenimiento) FROM gasto_por_vehiculo
)
ORDER BY GastoMantenimiento DESC;""",
    },
    "10. Conversión de reservas a contratos": {
        "desc": (
            "Agrupa las reservas por estado y calcula cuántas reservas distintas "
            "llegaron a tener al menos un contrato formalizado. El porcentaje "
            "se calcula sobre el total de reservas de cada estado."
        ),
        "sql": """SELECT
    r.Estado AS EstadoReserva,
    COUNT(DISTINCT r.CodReserva) AS CantidadReservas,
    COUNT(DISTINCT CASE
        WHEN c.NumContrato IS NOT NULL THEN r.CodReserva
    END) AS ReservasConContrato,
    ROUND(
        100.0 * COUNT(DISTINCT CASE
            WHEN c.NumContrato IS NOT NULL THEN r.CodReserva
        END) / NULLIF(COUNT(DISTINCT r.CodReserva), 0),
        2
    ) AS PorcentajeConversion
FROM RESERVA AS r
LEFT JOIN ALQUILER_EFECTIVO AS ae ON ae.CodReserva = r.CodReserva
LEFT JOIN CONTRATO AS c ON c.IdAlquiler = ae.IdAlquiler
GROUP BY r.Estado
ORDER BY CantidadReservas DESC;""",
    },
}


# -----------------------------------------------------------------------------
# 3. VALIDACIÓN Y EJECUCIÓN SEGURA DE CONSULTAS SQL
# -----------------------------------------------------------------------------


def remove_leading_sql_comments(sql_text):
    """Retira comentarios ubicados antes de la primera instrucción SQL."""
    text = sql_text.lstrip().lstrip("\ufeff")
    while text:
        text = text.lstrip()
        if text.startswith("--"):
            newline = text.find("\n")
            if newline == -1:
                return ""
            text = text[newline + 1 :]
        elif text.startswith("/*"):
            end_comment = text.find("*/", 2)
            if end_comment == -1:
                raise ValueError("Se encontró un comentario SQL sin cierre (*/).")
            text = text[end_comment + 2 :]
        else:
            break
    return text


def execute_query_securely(sql_query):
    """Ejecuta una única consulta SELECT/WITH en una conexión de solo lectura.

    Retorna (DataFrame, error, truncado). El autorizador SQLite constituye
    la barrera principal; la validación inicial mejora los mensajes de error.
    """
    if not isinstance(sql_query, str) or not sql_query.strip():
        return None, "Escribe una consulta SQL antes de ejecutarla.", False

    try:
        sql_start = remove_leading_sql_comments(sql_query)
    except ValueError as exc:
        return None, str(exc), False

    if not re.match(r"^(SELECT|WITH)\b", sql_start, flags=re.IGNORECASE):
        return (
            None,
            "Solo se permiten consultas que comiencen con SELECT o WITH.",
            False,
        )

    connection = None
    try:
        connection = get_read_only_connection(use_authorizer=True)
        # execute() acepta una sola sentencia y rechaza secuencias múltiples.
        cursor = connection.execute(sql_query)

        if cursor.description is None:
            return None, "La instrucción no produjo un conjunto de resultados.", False

        columns = [description[0] for description in cursor.description]
        rows = cursor.fetchmany(MAX_RESULT_ROWS + 1)
        was_truncated = len(rows) > MAX_RESULT_ROWS
        if was_truncated:
            rows = rows[:MAX_RESULT_ROWS]

        dataframe = pd.DataFrame.from_records(rows, columns=columns)
        return dataframe, None, was_truncated

    except sqlite3.DatabaseError as exc:
        message = str(exc)
        if "not authorized" in message.lower() or "authorization denied" in message.lower():
            message = (
                "SQLite rechazó la operación. El consultor solo admite lectura: "
                "SELECT, WITH, JOIN, subconsultas y funciones de consulta."
            )
        elif "one statement at a time" in message.lower():
            message = "Ejecuta una sola sentencia SQL por vez."
        return None, f"Error SQL: {message}", False
    except Exception as exc:
        return None, f"No fue posible ejecutar la consulta: {exc}", False
    finally:
        if connection is not None:
            connection.close()


# -----------------------------------------------------------------------------
# 4. VARIABLES DE ESTADO Y ELEMENTOS DE INTERFAZ
# -----------------------------------------------------------------------------

CURRENT_DATAFRAME = None

html_header = """
<div style="background:#1e293b;color:white;padding:20px;border-radius:10px;margin-bottom:14px;font-family:Arial,sans-serif;">
  <h2 style="margin:0;text-align:center;font-size:22px;">
    SISTEMA INTERACTIVO DE CONSULTAS — ALQUILER DE VEHÍCULOS
  </h2>
  <p style="margin:8px 0 0;text-align:center;color:#cbd5e1;font-size:13px;">
    Exploración del esquema relacional 3FN y análisis de información empresarial
  </p>
</div>
"""

status_ok, status_msg = verificar_estado_sistema()
status_color = "#059669" if status_ok else "#dc2626"
panel_estado = widgets.HTML(
    value=(
        "<div style='background:#f1f5f9;border-left:5px solid "
        f"{status_color};padding:11px;border-radius:5px;margin-bottom:12px;'>"
        f"<strong>Estado de la conexión:</strong> {html.escape(status_msg)}"
        "</div>"
    )
)

# Salidas independientes
output_a = widgets.Output()
output_b = widgets.Output()
output_c = widgets.Output()
output_exportacion = widgets.Output()

# Pestaña A: consultas predefinidas
selector_predefinida = widgets.Dropdown(
    options=list(PREDEFINED_QUERIES.keys()),
    description="Consulta:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="min(100%, 680px)"),
)

ver_sql_checkbox = widgets.Checkbox(
    value=False,
    description="Mostrar sentencia SQL",
    style={"description_width": "initial"},
)

btn_ejecutar_predefinida = widgets.Button(
    description="Ejecutar consulta",
    button_style="primary",
    icon="play",
    disabled=not status_ok,
    layout=widgets.Layout(width="190px"),
)

info_box_predefinida = widgets.HTML()

# Pestaña B: editor SQL personalizado
editor_sql = widgets.Textarea(
    placeholder="Escribe una consulta SELECT o WITH. Ejemplo: SELECT * FROM SUCURSAL;",
    layout=widgets.Layout(width="100%", height="170px"),
)
btn_ejecutar_personalizada = widgets.Button(
    description="Ejecutar consulta",
    button_style="success",
    icon="play",
    disabled=not status_ok,
)
btn_limpiar_editor = widgets.Button(
    description="Limpiar editor",
    button_style="warning",
    icon="trash",
)
btn_mostrar_tablas = widgets.Button(
    description="Mostrar tablas",
    button_style="info",
    icon="table",
    disabled=not status_ok,
)

# Pestaña C: explorador de tablas
selector_tabla = widgets.Dropdown(
    options=EXPECTED_TABLES,
    description="Tabla:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="min(100%, 390px)"),
    disabled=not status_ok,
)

filas_preview_slider = widgets.IntSlider(
    value=5,
    min=1,
    max=50,
    step=1,
    description="Filas de vista previa:",
    style={"description_width": "initial"},
    continuous_update=False,
    disabled=not status_ok,
)

# Herramientas compartidas para el último resultado
btn_descargar_csv = widgets.Button(
    description="Descargar resultados CSV",
    button_style="info",
    icon="download",
    disabled=True,
)



# -----------------------------------------------------------------------------
# 5. FUNCIONES DE ESTADO, RESULTADOS Y DESCRIPCIÓN DE CONSULTAS
# -----------------------------------------------------------------------------


def invalidate_current_result():
    """Invalida los resultados anteriores para evitar exportar datos obsoletos."""
    global CURRENT_DATAFRAME
    CURRENT_DATAFRAME = None
    btn_descargar_csv.disabled = True


def mostrar_error(output_widget, message):
    invalidate_current_result()
    with output_widget:
        clear_output(wait=True)
        display(
            HTML(
                "<div style='padding:10px;border:1px solid #fecaca;"
                "background:#fef2f2;color:#991b1b;border-radius:5px;'>"
                f"{html.escape(str(message))}</div>"
            )
        )


def presentar_resultados(dataframe, output_widget, title, truncated=False):
    """Presenta los resultados y habilita su descarga en CSV."""
    global CURRENT_DATAFRAME

    invalidate_current_result()
    CURRENT_DATAFRAME = dataframe.copy()

    with output_widget:
        clear_output(wait=True)
        display(
            HTML(
                "<div style='margin:10px 0;font-family:Arial,sans-serif;'>"
                f"<h3 style='color:#1e3a8a;margin:0;'>{html.escape(title)}</h3>"
                f"<p style='color:#64748b;margin:5px 0;'>"
                f"Registros: <strong>{len(dataframe):,}</strong> | "
                f"Columnas: <strong>{len(dataframe.columns)}</strong></p>"
                "</div>"
            )
        )

        if dataframe.empty:
            display(
                HTML(
                    "<div style='padding:10px;color:#92400e;background:#fffbeb;"
                    "border-radius:5px;'>La consulta se ejecutó correctamente, "
                    "pero no devolvió registros.</div>"
                )
            )
        else:
            display(dataframe)

        if truncated:
            display(
                HTML(
                    f"<p style='color:#92400e;'>Se muestran como máximo "
                    f"{MAX_RESULT_ROWS:,} filas. Refina la consulta para obtener "
                    "un conjunto más específico.</p>"
                )
            )

    # Una consulta válida, aunque no devuelva filas, puede descargarse con encabezados.
    btn_descargar_csv.disabled = False



def actualizar_info_predefinida(change=None):
    query_name = selector_predefinida.value
    query_data = PREDEFINED_QUERIES[query_name]
    visible_sql = ""

    if ver_sql_checkbox.value:
        visible_sql = (
            "<pre style='white-space:pre-wrap;overflow-x:auto;max-height:260px;"
            "padding:12px;background:#f8fafc;border:1px solid #e2e8f0;"
            "border-radius:5px;font-size:12px;'>"
            f"{html.escape(query_data['sql'])}</pre>"
        )

    info_box_predefinida.value = (
        "<div style='margin:10px 0;font-family:Arial,sans-serif;'>"
        f"<p><strong>Objetivo:</strong> {html.escape(query_data['desc'])}</p>"
        f"{visible_sql}</div>"
    )


# -----------------------------------------------------------------------------
# 6. EVENTOS DE LA INTERFAZ
# -----------------------------------------------------------------------------


def handler_ejecutar_predefinida(_button):
    current_status, current_message = verificar_estado_sistema()
    if not current_status:
        mostrar_error(output_a, current_message)
        return

    query_name = selector_predefinida.value
    query_sql = PREDEFINED_QUERIES[query_name]["sql"]
    dataframe, error, truncated = execute_query_securely(query_sql)

    if error:
        mostrar_error(output_a, error)
    else:
        presentar_resultados(dataframe, output_a, query_name, truncated)


def handler_ejecutar_personalizada(_button):
    query_sql = editor_sql.value.strip()
    if not query_sql:
        mostrar_error(output_b, "Escribe una consulta SQL antes de ejecutarla.")
        return

    dataframe, error, truncated = execute_query_securely(query_sql)
    if error:
        mostrar_error(output_b, error)
    else:
        presentar_resultados(
            dataframe,
            output_b,
            "Resultado de consulta personalizada",
            truncated,
        )


def handler_limpiar_editor(_button):
    editor_sql.value = ""
    with output_b:
        clear_output(wait=True)
    invalidate_current_result()


def handler_mostrar_tablas(_button):
    connection = None
    try:
        connection = get_read_only_connection()
        rows = connection.execute(
            "SELECT name FROM sqlite_master "
            "WHERE type='table' AND name NOT LIKE 'sqlite_%' ORDER BY name;"
        ).fetchall()
        table_names = [row[0] for row in rows]
        with output_b:
            clear_output(wait=True)
            display(
                HTML(
                    "<div style='padding:12px;background:#f0fdf4;"
                    "border:1px solid #bbf7d0;border-radius:5px;'>"
                    f"<strong>Tablas registradas ({len(table_names)}):</strong><br>"
                    f"<code>{html.escape(', '.join(table_names))}</code></div>"
                )
            )
    except Exception as exc:
        mostrar_error(output_b, str(exc))
    finally:
        if connection is not None:
            connection.close()


def quote_identifier(identifier):
    """Escapa un identificador SQL. Se usa junto con una lista blanca."""
    return '"' + str(identifier).replace('"', '""') + '"'


def handler_explorar_tabla(change=None):
    table_name = selector_tabla.value
    row_limit = int(filas_preview_slider.value)

    # El selector se alimenta únicamente de EXPECTED_TABLES.
    if table_name not in EXPECTED_TABLES:
        mostrar_error(output_c, "La tabla seleccionada no pertenece al esquema esperado.")
        return

    connection = None
    try:
        connection = get_read_only_connection()
        safe_table = quote_identifier(table_name)

        total_rows = connection.execute(
            f"SELECT COUNT(*) FROM {safe_table}"
        ).fetchone()[0]

        column_info = connection.execute(
            f"PRAGMA table_info({safe_table})"
        ).fetchall()
        metadata = pd.DataFrame(
            [tuple(row) for row in column_info],
            columns=["ID", "Columna", "Tipo de dato", "No nulo", "Valor por defecto", "Orden PK"],
        )
        if metadata.empty:
            raise RuntimeError("No se encontró información de columnas para esta tabla.")

        # En SQLite, el valor PK es 1, 2, ... para claves primarias compuestas.
        metadata["Clave primaria"] = metadata["Orden PK"].apply(
            lambda value: f"Sí (posición {value})" if value > 0 else "—"
        )
        metadata["No nulo"] = metadata["No nulo"].apply(
            lambda value: "Sí" if value else "No"
        )

        foreign_keys = connection.execute(
            f"PRAGMA foreign_key_list({safe_table})"
        ).fetchall()
        preview = pd.read_sql_query(
            f"SELECT * FROM {safe_table} LIMIT {row_limit};", connection
        )

        with output_c:
            clear_output(wait=True)
            display(
                HTML(
                    "<div style='margin-top:8px;'>"
                    f"<h3 style='color:#1e293b;'>Estructura de {html.escape(table_name)}</h3>"
                    f"<p><strong>Total de registros:</strong> {total_rows:,}</p>"
                    "<h4>Columnas, tipos y claves primarias</h4></div>"
                )
            )
            display(
                metadata[["Columna", "Tipo de dato", "No nulo", "Clave primaria", "Valor por defecto"]]
            )

            display(HTML("<h4>Claves foráneas</h4>"))
            if foreign_keys:
                fk_records = []
                for fk in foreign_keys:
                    fk_records.append(
                        {
                            "Columna local": fk[3],
                            "Tabla referenciada": fk[2],
                            "Columna referenciada": fk[4],
                            "Regla ON UPDATE": fk[5],
                            "Regla ON DELETE": fk[6],
                        }
                    )
                display(pd.DataFrame(fk_records))
            else:
                display(HTML("<p style='color:#64748b;'>No tiene claves foráneas declaradas.</p>"))

            display(HTML(f"<h4>Vista previa (primeras {row_limit} filas)</h4>"))
            display(preview)

    except Exception as exc:
        mostrar_error(output_c, f"No fue posible explorar la tabla: {exc}")
    finally:
        if connection is not None:
            connection.close()


# -----------------------------------------------------------------------------
# 7. DESCARGA CSV Y GRÁFICOS DEL ÚLTIMO RESULTADO
# -----------------------------------------------------------------------------


def handler_descargar_csv(_button):
    if CURRENT_DATAFRAME is None:
        return
    if colab_files is None:
        with output_exportacion:
            clear_output(wait=True)
            display(HTML("<p>La descarga directa está disponible dentro de Google Colab.</p>"))
        return

    filename = "reporte_consulta_alquiler_" + datetime.now().strftime("%Y%m%d_%H%M%S") + ".csv"
    CURRENT_DATAFRAME.to_csv(
        filename,
        index=False,
        encoding="utf-8-sig",
    )
    colab_files.download(filename)


# -----------------------------------------------------------------------------
# 8. COMPOSICIÓN DE LA INTERFAZ
# -----------------------------------------------------------------------------

selector_predefinida.observe(actualizar_info_predefinida, names="value")
ver_sql_checkbox.observe(actualizar_info_predefinida, names="value")
actualizar_info_predefinida()

selector_tabla.observe(handler_explorar_tabla, names="value")
filas_preview_slider.observe(handler_explorar_tabla, names="value")

btn_ejecutar_predefinida.on_click(handler_ejecutar_predefinida)
btn_ejecutar_personalizada.on_click(handler_ejecutar_personalizada)
btn_limpiar_editor.on_click(handler_limpiar_editor)
btn_mostrar_tablas.on_click(handler_mostrar_tablas)
btn_descargar_csv.on_click(handler_descargar_csv)

layout_a = widgets.VBox(
    [
        widgets.HTML("<h4>Consultas estratégicas predefinidas</h4>"),
        widgets.HBox([selector_predefinida, ver_sql_checkbox]),
        info_box_predefinida,
        btn_ejecutar_predefinida,
        widgets.HTML("<hr>"),
        output_a,
    ],
    layout=widgets.Layout(width="100%"),
)

layout_c = widgets.VBox(
    [
        widgets.HTML("<h4>Explorador del esquema relacional</h4>"),
        widgets.HBox([selector_tabla, filas_preview_slider]),
        widgets.HTML("<hr>"),
        output_c,
    ],
    layout=widgets.Layout(width="100%"),
)

# El editor SQL se mostrará directamente, sin quedar anidado en VBox/HBox.
# Esta composición reduce posibles problemas de foco en Google Colab.

panel_exportacion = widgets.VBox(
    [
        widgets.HTML(
            "<h3 style='border-bottom:2px solid #cbd5e1;padding-bottom:7px;'>"
            "4. Exportar resultados</h3>"
            "<p style='color:#64748b;font-size:13px;'>"
            "Descarga en CSV el resultado de la última consulta ejecutada.</p>"
        ),
        btn_descargar_csv,
        output_exportacion,
    ],
    layout=widgets.Layout(width="100%", padding="12px"),
)

# -----------------------------------------------------------------------------
# 9. MOSTRAR INTERFAZ
# -----------------------------------------------------------------------------

display(HTML(html_header))
display(panel_estado)
display(HTML(
    "<div style='padding:10px 12px;margin:8px 0 16px;background:#eff6ff;"
    "border-left:4px solid #3b82f6;border-radius:5px;'>"
    "<strong>¿Dónde consultar?</strong> Esta franja solo informa el estado de la conexión; "
    "no es un espacio para escribir SQL. Usa <strong>1. Consultas predefinidas</strong> "
    "para ejecutar análisis ya preparados o <strong>2. SQL personalizado</strong> "
    "para escribir tus propias consultas. La sección <strong>3. Explorador de tablas</strong> "
    "permite revisar columnas, claves y registros sin escribir SQL."
    "</div>"
))

display(HTML("<h3 style='border-bottom:2px solid #cbd5e1;padding-bottom:7px;'>1. Consultas predefinidas</h3>"))
display(layout_a)

display(HTML("<h3 style='border-bottom:2px solid #cbd5e1;padding-bottom:7px;'>2. SQL personalizado</h3>"))
display(HTML(
    "<h4>Consultor SQL personalizado</h4>"
    "<p>Se admiten consultas SELECT y WITH. "
    "Las operaciones que modifican la base de datos son rechazadas.</p>"
))
# Importante: editor_sql se presenta como widget raíz, no dentro de un contenedor.
display(editor_sql)
display(widgets.HBox([btn_ejecutar_personalizada, btn_limpiar_editor, btn_mostrar_tablas]))
display(HTML("<hr>"))
display(output_b)

display(HTML("<h3 style='border-bottom:2px solid #cbd5e1;padding-bottom:7px;'>3. Explorador de tablas</h3>"))
display(layout_c)
display(panel_exportacion)

if status_ok:
    handler_explorar_tabla()
else:
    with output_c:
        display(
            HTML(
                "<p style='color:#991b1b;'>"
                "El explorador está deshabilitado hasta que la base de datos esté disponible.</p>"
            )
        )

print(
    "Módulo interactivo cargado. La base de datos se consulta en modo de solo lectura; "
    "la conexión original del cuaderno no se modifica."
)
````
